### 2.2 Renombrar columnas

**🎯Objetivo:**
Estandarizar los nombres de columnas para evitar errores y facilitar la unión de los datasets.

**Instrucciones:**

- Cambia los nombres de las columnas para que tengan el formato `snake_case`.
    - `Country` → `country`
    - `UpdateTimeUTC` → `update_time_utc`
- Verifica que los cambios se hayan aplicado correctamente usando `.columns`.


```python
# Estandarizar los nombres de las columnas de traffic

#Intenté instalar janitor para facilitar la tarea, pero no está permitido #!pip install pyjanitor
#import janitor
#traffic = traffic.clean_names()

#cortesía Gemini
import re

def to_snake_acronyms(name):
    # Separa minúsculas seguidas de mayúsculas (ej: UpdateTime -> Update_Time)
    s = re.sub(r'([a-z0-9])([A-Z])', r'\1_\2', name)
    # Reemplaza espacios, guiones y caracteres especiales por guiones bajos
    s = re.sub(r'[\s\-]+', '_', s)
    # Limpia guiones bajos duplicados y convierte a minúsculas
    return re.sub(r'_+', '_', s).lower()

# Aplica la regla a tu DataFrame traffic
traffic.columns = [to_snake_acronyms(col) for col in traffic.columns]

#Este lo hice manual
traffic = traffic.rename(columns={'update_time_utcweek_ago':'update_time_utc_week_ago',
                                  'travel_time_live_per10_kms_mins':'travel_time_live_per_10_kms_mins',
                                  'travel_time_historic_per10_kms_mins':'travel_time_historic_per_10_kms_mins'
                                 })
# verificar cambios
traffic.columns
```

```python
# Estandarizar los nombres de las columnas de eco

#cortesía Gemini
import re

def to_snake_acronyms(text):
    # Convertir caracteres especiales Unicode a texto normal
    text = text.replace('μ', 'u').replace('³', '3').replace('/', '_').replace('%', 'pct')
    text = re.sub(r'(\d+)\.(\d+)', r'\1\2', text)
    text = re.sub(r'[^\w\s\-]', '', text)
    text = re.sub(r'[\s\-]+', '_', text)
    text = re.sub(r'_+', '_', text)
    return text.strip('_').lower()

# Aplica nuevamente sobre tu DataFrame eco
eco.columns = [to_snake_acronyms(col) for col in eco.columns]

# verificar cambios
eco.columns
```

```python
# Calcular los  promedios de trafico por ciudad, país y año
# Lista de métricas de tráfico requeridas
metrics = [
    'jams_delay', 
    'traffic_index_live', 
    'jams_length_in_kms', #estoy usando el nombre que yo le asigné para no dañar la continuidad del código
    'jams_count', 
    'mins_delay', 
    'travel_time_live_per_10_kms_mins', #estoy usando el nombre que yo le asigné para no dañar la continuidad del código
    'travel_time_historic_per_10_kms_mins' #estoy usando el nombre que yo le asigné para no dañar la continuidad del código
]

traffic_city_year_2024 = (
    traffic_2024
    .groupby(['city', 'country', 'year'], as_index=False)[metrics]
    .mean()
)

# Mostrar resultado
traffic_city_year_2024.head()
```

### 5.1 Unir tráfico (tabla principal) con indicadores económicos

**🎯Objetivo:**
Combinar la información de tráfico y economía en un solo DataFrame para analizar cómo las condiciones económicas se relacionan con la movilidad urbana.

**Instrucciones**
- Selecciona solo las **columnas relevantes** de cada dataset (por ejemplo, variables clave de tráfico y de economía).
- Usa `.copy()` al crear subconjuntos para evitar modificar el dataset original.
- Une ambos DataFrames y define como **claves de unión** a `city` y `year`.

- Guarda el resultado en una nueva variable llamada `merged` y muestra las primeras 5 filas.


<details>
<summary>Haz clic para ver la pista</summary>
Aplica una unión de tipo "inner" para mantener las ciudades y años presentes en ambos datasets.

```python

# Seleccionar columnas clave de tráfico y economía
left_cols = ['city','country','year','jams_delay','traffic_index_live',
             'jams_length_in_kms','jams_count','mins_delay',
             'travel_time_live_per_10_kms_mins','travel_time_historic_per_10_kms_mins'] #estoy usando el nombre que yo le asigné para no dañar la continuidad del código

right_cols = ['city','year','city_gdp_capita','unemployment_pct','pm25_ug_m3','population'] #estoy usando el nombre que yo le asigné para no dañar la continuidad del código

# Usar .copy() para crear los dos nuevos datasets reducidos
traffic_2024_small = traffic_city_year_2024[left_cols].copy()
eco_2024_small = eco_2024[right_cols].copy()

# Unir datasets
merged = pd.merge(traffic_2024_small, eco_2024_small, on=['city','year'],how='inner')


# Mostrar las primeras 5 filas
merged.head()
```
